```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

time_labels = ["2020年11月", "2020年12月", "2021年1月", "2021年2月", "2021年3月", "2021年4月", "2021年5月"]
apply_volume = [64, 208, 273, 282, 271, 269, 230]
month_on_month = [np.nan, 2.25, 0.313, 0.033, -0.039, -0.007, -0.145]
percent_text = ["", "225.0%", "31.3%", "3.3%", "-3.9%", "-0.7%", "-14.5%"]

x = np.arange(len(time_labels))
width = 0.5

fig, ax1 = plt.subplots(figsize=(12,6), dpi=120)

bars = ax1.bar(x, apply_volume, width, color="#4472c4", label="贷款申请量")
ax1.set_xlabel("")
ax1.set_ylabel("贷款申请量", fontsize=11)
ax1.set_ylim(0, 320)
ax1.set_xticks(x)
ax1.set_xticklabels(time_labels)

for bar in bars:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 6,
             f"{int(height)}", ha="center", va="bottom", fontsize=14)

ax2 = ax1.twinx()
line = ax2.plot(x, month_on_month, marker="o", color="#ed7d31", linewidth=3, markersize=8, label="环比变化率")
ax2.set_ylabel("环比变化率", fontsize=11)
ax2.set_ylim(-0.5, 2.5)
# 标注百分比文字
for i, txt in enumerate(percent_text):
    if txt != "":
        ax2.text(x[i], month_on_month[i]+0.08, txt, color="red", fontsize=14, ha="center")

# 图例合并
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper center", ncol=2, frameon=False)

plt.title("图 1-1 贷款申请月度变化趋势", fontsize=14, pad=20)
plt.tight_layout()
plt.savefig("loan_apply_trend.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = ["SimHei"]
plt.rcParams["axes.unicode_minus"] = False

# 数据
time_labels = ["2020年11月", "2020年12月", "2021年1月", "2021年2月", "2021年3月", "2021年4月", "2021年5月"]
manager =     [21, 99, 138, 132, 142, 130, 107]  # 理财经理（柜台）蓝色
mobile =      [37, 95, 118, 135, 113, 122, 103]  # 手机银行橙色
self_machine =[6, 14, 17, 15, 16, 17, 20]        # 自助机黄色

x = np.arange(len(time_labels))
width = 0.6

fig, ax = plt.subplots(figsize=(12,7), dpi=120)

# 绘制堆叠柱状图
bar1 = ax.bar(x, manager, width, label="理财经理（柜台）", color="#4472c4")
bar2 = ax.bar(x, mobile, width, bottom=manager, label="手机银行", color="#ed7d31")
bar3 = ax.bar(x, self_machine, width, bottom=np.array(manager)+np.array(mobile), label="自助机", color="#ffc000")

# 在每个分段柱子上标注数字
for i in range(len(x)):
    # 理财经理
    ax.text(x[i], manager[i]/2, str(manager[i]), ha="center", va="center", fontsize=14)
    # 手机银行
    ax.text(x[i], manager[i] + mobile[i]/2, str(mobile[i]), ha="center", va="center", fontsize=14)
    # 自助机
    ax.text(x[i], manager[i]+mobile[i] + self_machine[i]/2, str(self_machine[i]), ha="center", va="center", fontsize=14)

ax.set_xticks(x)
ax.set_xticklabels(time_labels)
ax.set_ylim(0, 160)
ax.legend(loc="upper center", ncol=3, frameon=False)
ax.set_title("图 1-2 各渠道贷款申请构成月度变化趋势", fontsize=14, pad=20)

plt.tight_layout()
plt.savefig("loan_channel_stack_bar.png", dpi=150, bbox_inches="tight")
plt.show()
```
